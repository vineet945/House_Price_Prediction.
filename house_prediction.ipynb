## Handling Categorical Missing Values

```python
df_cat_clm = df_drop_clm.select_dtypes(include=['object'])
df_cat_clm.isnull().sum()[df_cat_clm.isnull().sum() > 0]
```

```python
# Fill remaining categorical nulls with the mode (most frequent value) of each column
df_cat_clm2 = df_cat_clm.copy()
for col in df_cat_clm2.columns[df_cat_clm2.isnull().any()]:
    df_cat_clm2[col] = df_cat_clm2[col].fillna(df_cat_clm2[col].mode()[0])

df_cat_clm2.isnull().sum().sum()  # should be 0
```

## Final Data Assembly for Modeling

```python
# Numeric columns: same as df_num_clm, but with all originally-null columns (l1) median-imputed
df_num_final = df_num_clm.copy()
for i in l1:
    df_num_final[i] = df_num_final[i].fillna(df_num_final[i].median())

df_num_final.isnull().sum().sum()  # should be 0
```

```python
# Combine fully-imputed numeric + categorical columns
df_clean = pd.concat([df_num_final, df_cat_clm2], axis=1)
print(df_clean.shape)
print('Remaining nulls:', df_clean.isnull().sum().sum())
```

```python
# One-hot encode categorical columns
X = pd.get_dummies(df_clean.drop(columns=['SalePrice']), drop_first=True)
y = df_clean['SalePrice']
X.shape, y.shape
```

```python
from sklearn.model_selection import train_test_split

X_train, X_valid, Y_train, Y_valid = train_test_split(
    X, y, test_size=0.2, random_state=0
)
print(X_train.shape, X_valid.shape)
```

## Modeling

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

def evaluate(name, y_true, y_pred):
    mae = mean_absolute_error(y_true, y_pred)
    rmse = mean_squared_error(y_true, y_pred) ** 0.5
    r2 = r2_score(y_true, y_pred)
    print(f'{name}')
    print(f'  MAE  : {mae:,.0f}')
    print(f'  RMSE : {rmse:,.0f}')
    print(f'  R2   : {r2:.4f}')
    return {'Model': name, 'MAE': mae, 'RMSE': rmse, 'R2': r2}
```

```python
from sklearn.linear_model import LinearRegression

lr = LinearRegression()
lr.fit(X_train, Y_train)
Y_pred_lr = lr.predict(X_valid)

res_lr = evaluate('Linear Regression', Y_valid, Y_pred_lr)
```

```python
from sklearn.ensemble import RandomForestRegressor

rf = RandomForestRegressor(n_estimators=200, random_state=0)
rf.fit(X_train, Y_train)
Y_pred_rf = rf.predict(X_valid)

res_rf = evaluate('Random Forest', Y_valid, Y_pred_rf)
```

```python
from sklearn.ensemble import GradientBoostingRegressor

gbr = GradientBoostingRegressor(n_estimators=200, learning_rate=0.05, max_depth=3, random_state=0)
gbr.fit(X_train, Y_train)
Y_pred_gbr = gbr.predict(X_valid)

res_gbr = evaluate('Gradient Boosting', Y_valid, Y_pred_gbr)
```

## Model Comparison

```python
results = pd.DataFrame([res_lr, res_rf, res_gbr]).sort_values('R2', ascending=False)
results
```

```python
plt.figure(figsize=(7,5))
sns.barplot(x='Model', y='R2', data=results)
plt.title('Model Comparison - R2 Score (higher is better)')
plt.ylim(0, 1)
plt.show()
```

## Best Model: Feature Importance & Prediction Quality

```python
best_name = results.iloc[0]['Model']
best_model = {'Linear Regression': lr, 'Random Forest': rf, 'Gradient Boosting': gbr}[best_name]
best_pred = {'Linear Regression': Y_pred_lr, 'Random Forest': Y_pred_rf, 'Gradient Boosting': Y_pred_gbr}[best_name]
print('Best model:', best_name)
```

```python
if hasattr(best_model, 'feature_importances_'):
    importances = pd.Series(best_model.feature_importances_, index=X.columns).sort_values(ascending=False).head(15)
    plt.figure(figsize=(8,6))
    importances.plot(kind='barh')
    plt.title(f'{best_name} - Top 15 Feature Importances')
    plt.gca().invert_yaxis()
    plt.show()
else:
    coefs = pd.Series(best_model.coef_, index=X.columns).sort_values(key=abs, ascending=False).head(15)
    print(coefs)
```

```python
plt.figure(figsize=(7,7))
plt.scatter(Y_valid, best_pred, alpha=0.5)
plt.plot([Y_valid.min(), Y_valid.max()], [Y_valid.min(), Y_valid.max()], 'r--')
plt.xlabel('Actual SalePrice')
plt.ylabel('Predicted SalePrice')
plt.title(f'{best_name}: Actual vs Predicted SalePrice')
plt.show()
```
